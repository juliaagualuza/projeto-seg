# Análise exploratória — Criminalidade no Estado do Rio de Janeiro

Este notebook carrega o CSV gerado pelo ETL e realiza verificações e análises exploratórias. Execute primeiro `python src/etl.py`.


```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

arquivo = Path('../data/processed/isp_rj_processed.csv')
df = pd.read_csv(arquivo)
df.head()
```

```python
print('Dimensões:', df.shape)
print('Duplicatas:', df.duplicated().sum())
display(df.isna().sum().sort_values(ascending=False).head(15))
```

```python
if {'ano', 'registro_ocorrencias'}.issubset(df.columns):
    anual = (df.groupby('ano', as_index=False)['registro_ocorrencias']
               .sum(min_count=1).sort_values('ano'))
    display(anual)
    # 2026 é parcial (jan–set na versão analisada); destaque-o separadamente.
    completo = anual[anual['ano'] < 2026]
    ax = completo.plot(x='ano', y='registro_ocorrencias', kind='line', marker='o', legend=False)
    ax.set_title('Registros de ocorrências por ano (2014–2025)')
    ax.set_xlabel('Ano')
    ax.set_ylabel('Registros')
    plt.tight_layout()
    plt.show()
else:
    print('As colunas ano e registro_ocorrencias não foram encontradas.')
```

```python
if {'fmun', 'registro_ocorrencias'}.issubset(df.columns):
    municipios = (df.groupby('fmun', as_index=False)['registro_ocorrencias']
                    .sum(min_count=1)
                    .sort_values('registro_ocorrencias', ascending=False))
    display(municipios.head(10))
```
